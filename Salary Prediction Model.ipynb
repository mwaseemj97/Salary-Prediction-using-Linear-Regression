# Salary Prediction Model

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, precision_score, f1_score, confusion_matrix, classification_report
```

```python
df = pd.read_csv('Salary_Data.csv')

df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
X = df.drop(columns=['Salary'])
y = df['Salary']
```

```python
X.head()
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 90)
```

```python
X_train
```

```python
len(X_train)/len(df)*100
```

```python
len(X_test)/len(df)*100
```

```python
model = LinearRegression()
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

y_pred
```

```python
%matplotlib inline

plt.scatter(x=df['YearsExperience'], y=df['Salary'], color = 'r', marker = '+')
plt.title('Salary Prediction Plot')
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.savefig('Salary')
plt.show()
```

```python
%matplotlib inline


plt.title('Salary Prediction Plot')
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.scatter(x=df['YearsExperience'], y=df['Salary'], color = 'r', marker = '+')
plt.plot(df['YearsExperience'], model.predict(df[['YearsExperience']]), color='blue')
plt.savefig('Salary_2')
```
